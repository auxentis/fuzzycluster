```python
output_file = 'human_pseudo_genome.fna'
```

```python
def fasta_reader(infile):
    """
    Parsing fasta
    """
    with open(infile) as handle:
        for record in FastaIterator(handle):
            yield record
```

```python
def calculate_frequencies(sequence, letters):
    """
    Calculate the frequency of each letter in the provided sequences.

    Parameters:
    - sequences (list of str): The list of sequences.

    Returns:
    - list: The frequencies of each letter in the order of letters.
    """
    count = Counter(sequence)
    total = sum(count.values())
    frequencies = [count[letter] / total for letter in letters]
    return frequencies
```

```python
def generate_random_sequence(length, letters, frequencies):
    """
    Generate a random sequence of a given length from a specified set of letters
    based on their respective frequencies.

    Parameters:
    - length (int): The length of the sequence to generate.
    - letters (list or str): The set of letters to choose from.
    - frequencies (list or np.array): The frequency of each letter.

    Returns:
    - np.array: A NumPy array containing the random sequence.
    """
    sequence = np.random.choice(letters, size=length, p=frequencies)
    return sequence
```

```python
def create_chromosome(description, sequence):
    """
    Create a SeqRecord for a chromosome.

    Parameters:
    - description (str): The description to include in the FASTA header.
    - sequence (str): The sequence of the chromosome.

    Returns:
    - SeqRecord: A SeqRecord object representing the chromosome.
    """
    return SeqRecord(Seq(sequence), id=description, description="pseudo")
```

```python
def save_genome_to_file(chromosomes, file_path):
    """
    Save the list of SeqRecord objects to a file in FASTA format.

    Parameters:
    - chromosomes (list of SeqRecord): The list of SeqRecord objects.
    - file_path (str): The path to the file where the genome will be saved.
    """
    SeqIO.write(chromosomes, file_path, "fasta")
```

```python
def artificial_genome_multifasta(filename): 
    chromosomes = []   

    for entry in fasta_reader(filename):
        seqname = entry.id
        seq = entry.seq
        description = entry.description
        print(description)
        myseq = str(seq).replace(' ', '').upper()
        letters = ['A', 'C', 'G', 'T']
        length = len(myseq)

        
        letter_frequencies = calculate_frequencies(myseq, letters)
        pseudo_generated = generate_random_sequence(length, letters, letter_frequencies)
        pseudo_seq = ''.join(pseudo_generated)

        
        pseudo_description = f"pseudo_{description}"
        pseudo_chromosome = create_chromosome(pseudo_description, pseudo_seq)
        chromosomes.append(pseudo_chromosome)
        result = save_genome_to_file(chromosomes, output_file)
    
    return result
```

```python
artificial_genome = artificial_genome_multifasta(filename)
```
